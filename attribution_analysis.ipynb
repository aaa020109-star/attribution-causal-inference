## 1. 데이터 로드 + touch_set 구성 (정제 단계,로그 포함)

```python
import pandas as pd
import numpy as np

df = pd.read_csv('data/attribution_20260828_1534.csv', encoding='utf-8-sig')

channels = ['Email', 'Push', 'Pmax', 'Affiliate', 'PaidSearch']
visit_cols = {'Email': 'value1', 'Push': 'value3', 'Pmax': 'value5', 'Affiliate': 'value7', 'PaidSearch': 'value9'}
order_cols = {'Email':'value2','Push':'value4','Pmax':'value6','Affiliate':'value8','PaidSearch':'value10'}

df_clean = pd.DataFrame()
df_clean['visitor_id'] = df['value']
for ch in channels:
    df_clean[ch] = (df[visit_cols[ch]] > 0).astype(int)
df_clean['converted'] = (df[list(order_cols.values())].sum(axis=1) > 0).astype(int)

df_clean['touch_set'] = df_clean[channels].apply(
    lambda r: frozenset([c for c in channels if r[c] == 1]), axis =1 
)

#===== 로그 출력 =====
print(f" 전체 {len(df_clean)}명 정제 완료")
print(f"전체 전환율: {df_clean['converted'].mean():.2%}")
print(f"\n채널별 접촉자 수:")
print(df_clean[channels].sum())
print(f"\n관측된 조합 수: {df_clean['touch_set'].nunique()}개 (이론상 최대 {2**len(channels)-1}개")
print(df_clean.head())
```

## 2. 조합별 통계 집계

```python
coalition_stats = df_clean.groupby('touch_set').agg(
    n_users= ('converted', 'size'),
    n_conv=('converted', 'sum')
).reset_index()
coalition_stats['conv_rate'] = coalition_stats['n_conv'] / coalition_stats['n_users']

# 로그 출력
print(f"조합별 통계 집계 완료: {len(coalition_stats)}개 조합")
print(coalition_stats.sort_values('n_users', ascending=False).to_string(index=False))
```

## 3. 가치함수 v(S) + Shapley Value 계산 함수 정의

```python
from itertools import combinations
import math

def v_fixed(S, coalition_stats):
    S = frozenset(S)
    mask = coalition_stats['touch_set'].apply(lambda ts: ts.issubset(S))
    subset = coalition_stats[mask]
    return subset['n_conv'].sum() # 전환 건수(count), 비율 아님

def shapley_values_fixed(channels, coalition_stats, verbose=False):
    n = len(channels)
    phi = {c: 0.0 for c in channels}
    for i in channels:
        others = [c for c in channels if c != i]
        for r in range(len(others) + 1):
            for S in combinations(others, r):
                S = set(S)
                weight = (math.factorial(len(S)) * math.factorial(n - len(S) -1)) / math.factorial(n)
                marginal = v_fixed(S | {i}, coalition_stats) - v_fixed(S, coalition_stats)
                phi[i] += weight * marginal
        if verbose:
            print(f" '{i}' 계산 완료: {phi[i]:.2f}")
    return phi

print("v(S), shapley_values 함수 정의 완료")
```

## 4. 실제 데이터로 Shapley Value 계산 실행

```python
print("⏳ 수정된 Shapley Value 계산 중...")
shap_fixed = shapley_values_fixed(channels, coalition_stats, verbose=True)

total_conv = df_clean['converted'].sum()
print(f"\n검증: Shapley 값 합계 = {sum(shap_fixed.values()):.2f}, 실제 전체 전환자 수 = {total_conv}")
print("(이 두 값이 거의 같아야 정상입니다 — Shapley의 효율성(efficiency) 속성)")

shap_pct = {ch: v/sum(shap_fixed.values()) for ch, v in shap_fixed.items()}
print("\n✅ 최종 채널별 전환 기여도:")
for ch, val in sorted(shap_pct.items(), key=lambda x: -x[1]):
    print(f"  {ch}: {val:.1%}")
```

## 6. 부트스트랩으로 각 채널 기여도 신뢰구간 계산

```python
def bootstrap_shapley(df_clean, channels, n_bootstrap=50):
    results = {c: [] for c in channels}
    for i in range(n_bootstrap):
        sample = df_clean.sample(frac=1, replace=True)
        cs = sample.groupby('touch_set').agg(
            n_users=('converted', 'size'), n_conv=('converted', 'sum')
        ).reset_index()
        sv = shapley_values_fixed(channels, cs, verbose=False)
        total = sum(sv.values())
        for c in channels:
            results[c].append(sv[c] / total if total > 0 else 0)
        if (i+1) % 10 == 0:
            print(f"[{i+1}/{n_bootstrap}] 진행 중...")
    return results

print("부트스트랩 시작 (50회)")
boot_results = bootstrap_shapley(df_clean, channels, n_bootstrap=50)

print("\n채널별 기여도 95% 신뢰구간:")
for ch in channels:
    vals = boot_results[ch]
    mean = np.mean(vals)
    ci_low, ci_high = np.percentile(vals,[2.5, 97.5])
    print(f" {ch}: 평균 {mean:.1%}, 95% CI [{ci_low:.1%}, {ci_high:.1%}]")
```

## 7. Affiliate가 진짜 1등인지 paired bootstrap으로 재검증

```python
def bootstrap_shapley_difference(df_clean, channels, channel_a, channel_b, n_bootstrap=50):
    diffs = []
    for i in range(n_bootstrap):
        sample = df_clean.sample(frac=1, replace=True)
        cs = sample.groupby('touch_set').agg(
            n_users=('converted', 'size'), n_conv=('converted','sum')
        ).reset_index()
        sv = shapley_values_fixed(channels, cs, verbose=False)
        total = sum(sv.values())
        if total == 0:
            continue
        pct_a = sv[channel_a] / total
        pct_b = sv[channel_b] / total
        diffs.append(pct_a - pct_b)
    return diffs

other_channels = [ch for ch in channels if ch != 'Affiliate']
print("Affiliate vs 나머지 채널 차이 검증 중...\n")

for ch in other_channels:
    diffs = bootstrap_shapley_difference(df_clean, channels, 'Affiliate', ch, n_bootstrap=50)
    ci_low, ci_high = np.percentile(diffs, [2.5, 97.5])
    is_sig = ci_low > 0
    print(f"Affiliate - {ch}: 95% CI [{ci_low:.1%}, {ci_high:.1%}] > {'유의미' if is_sig else '유의미하지 않음'}")
```

## 11. v(S) 두 버전 비교 (count vs rate)

```python
def v_count(S, coalition_stats):
    S = frozenset(S)
    mask = coalition_stats['touch_set'].apply(lambda ts: ts.issubset(S))
    subset = coalition_stats[mask]
    return subset['n_conv'].sum()

def v_rate(S, coalition_stats):
    S = frozenset(S)
    mask = coalition_stats['touch_set'].apply(lambda ts: ts.issubset(S))
    subset = coalition_stats[mask]
    if subset['n_users'].sum() == 0:
        return 0.0
    return subset['n_conv'].sum() / subset['n_users'].sum()

def shapley_generic(channels, coalition_stats, v_func):
    n = len(channels)
    phi = {c: 0.0 for c in channels}
    for i in channels:
        others = [c for c in channels if c != i]
        for r in range(len(others) + 1):
            for S in combinations(others, r):
                S = set(S)
                weight = (math.factorial(len(S)) * math.factorial(n - len(S) - 1)) / math.factorial(n)
                marginal = v_func(S | {i}, coalition_stats) - v_func(S, coalition_stats)
                phi[i] += weight * marginal
    return phi

print("Count 기반 Shapley (기존 방식)...")
shap_count = shapley_generic(channels, coalition_stats, v_count)
total_count = sum(shap_count.values())

print("Rate 기반 Shapley (노출량 편향 완화)...")
shap_rate = shapley_generic(channels, coalition_stats, v_rate)
total_rate = sum(shap_rate.values())

print("\n 비교 결과:")
print(f"{'채널':<12}{'Count 기반':<15}{'Rate 기반':<15}{'접촉자수':<10}")
for ch in channels:
    n_touched = df_clean[ch].sum()
    print(f"{ch:<12}{shap_count[ch]/total_count:<15.1%}{shap_rate[ch]/total_rate:<15.1%}{n_touched:<10}")
```

## 18. IPW 보정된 v(S) 함수로 Shapley 재계산

```python
def compute_weighted_coalition_stats(df_clean, channels):
    # 각 방문자의 조합별 대표 가중치: 그 사람이 접촉한 채널들의 IPW 평균 사용
    def get_weight(row):
        touched = [ch for ch in channels if row[ch] == 1]
        if not touched:
            return 1.0
        return np.mean([row[f'ipw_{ch}'] for ch in touched])

    df_clean['combined_weight'] = df_clean.apply(get_weight, axis=1)

    stats = df_clean.groupby('touch_set').apply(
        lambda g: pd.Series({
            'n_users': g['combined_weight'].sum(),   # 가중합
            'n_conv': (g['combined_weight'] * g['converted']).sum()  # 가중 전환수
        })
    ).reset_index()
    return stats

print(" IPW 가중 coalition_stats 계산 중...")
weighted_coalition_stats = compute_weighted_coalition_stats(df_clean, channels)
print(weighted_coalition_stats.sort_values('n_users', ascending=False).head(10))
```

## 20. IPW 버전에 부트스트랩 적용(통계적 유의성 확인)

```python
def bootstrap_ipw_shapley(df_clean, channels, covariates, n_bootstrap=20, clip_upper=10):
    results = {c: [] for c in channels}
    for i in range(n_bootstrap):
        sample = df_clean.sample(frac=1, replace=True).reset_index(drop=True)

        # 성향점수 재학습
        for ch in channels:
            model = LogisticRegression()
            model.fit(sample[covariates], sample[ch])
            ps = model.predict_proba(sample[covariates])[:, 1]
            raw_w = np.where(sample[ch]==1, 1/ps, 1/(1-ps))
            sample[f'ipw_{ch}'] = np.clip(raw_w, None, clip_upper)

        def get_weight(row):
            touched = [ch for ch in channels if row[ch] == 1]
            if not touched:
                return 1.0
            return np.mean([row[f'ipw_{ch}'] for ch in touched])
        sample['combined_weight'] = sample.apply(get_weight, axis=1)

        sample['touch_set'] = sample[channels].apply(
            lambda r: frozenset([c for c in channels if r[c] == 1]), axis=1
        )
        cs = sample.groupby('touch_set').apply(
            lambda g: pd.Series({
                'n_users': g['combined_weight'].sum(),
                'n_conv': (g['combined_weight'] * g['converted']).sum()
            })
        ).reset_index()

        sv = shapley_generic(channels, cs, v_count)
        total = sum(sv.values())
        for c in channels:
            results[c].append(sv[c]/total if total > 0 else 0)

        if (i+1) % 5 == 0:
            print(f"  [{i+1}/{n_bootstrap}] 진행 중...")
    return results

print("⏳ IPW 부트스트랩 시작 (20회, 시간 소요될 수 있음)...")
ipw_boot_results = bootstrap_ipw_shapley(df_clean, channels, covariates, n_bootstrap=20)

print("\n✅ IPW 보정 채널별 기여도 95% 신뢰구간:")
for ch in channels:
    vals = ipw_boot_results[ch]
    mean = np.mean(vals)
    ci_low, ci_high = np.percentile(vals, [2.5, 97.5])
    print(f"  {ch}: 평균 {mean:.1%}, 95% CI [{ci_low:.1%}, {ci_high:.1%}]")
```

## Two-Proportion Z-test - 채널 쌍 간 전환율 직접 비교

```python
from statsmodels.stats.proportion import proportions_ztest

print(" 채널 간 전환율 Two-Proportion Z-test\n")

conv_counts = {}
conv_totals = {}
for ch in channels:
    sub = df_clean[df_clean[ch] == 1]
    conv_counts[ch] = sub['converted'].sum()
    conv_totals[ch] = len(sub)

pairs = [('Affiliate', 'PaidSearch'), ('PaidSearch', 'Pmax'), ('Pmax', 'Email'), ('Email', 'Push')]
for a, b in pairs:
    count = [conv_counts[a], conv_counts[b]]
    nobs = [conv_counts[a], conv_totals[b]]
    z_stat, p_val = proportions_ztest(count, nobs)
    rate_a = conv_counts[a]/conv_totals[a]
    rate_b = conv_counts[b]/conv_totals[b]
    print(f"{a}({rate_a:.2%}) vs {b}({rate_b:.2%}): Z={z_stat:.2f}, p={p_val:.6f} -> {'유의미' if p_val < 0.05 else '유의미하지 않음'}")
```

## IPW 균형 진단(Covariate Balance Check)

```python
def check_balance(df_clean, channels, covariates):
    print(f"{'채널':<12}{'변수':<18}{'가중치 전 SMD':<15}{'가중치 후 SMD':<15}")
    for ch in channels:
        treated = df_clean[df_clean[ch]==1]
        control = df_clean[df_clean[ch]==0]
        w_treated = treated[f'ipw_{ch}']
        w_control = control[f'ipw_{ch}']
        for cov in covariates:
            # 가중치 전 SMD (표준화 평균 차이)
            smd_before = (treated[cov].mean() - control[cov].mean()) / df_clean[cov].std()
            # 가중치 후 SMD
            wm_t = np.average(treated[cov], weights=w_treated)
            wm_c = np.average(control[cov], weights=w_control)
            smd_after = (wm_t - wm_c) / df_clean[cov].std()
            print(f"{ch:<12}{cov:<18}{smd_before:<15.3f}{smd_after:<15.3f}")

check_balance(df_clean, channels, covariates)
```

## Trimming 민감도 분석

```python
def run_ipw_shapley(df_clean, channels, covariates, clip_upper=None):
    df_tmp = df_clean.copy()
    for ch in channels:
        model = LogisticRegression()
        model.fit(df_tmp[covariates], df_tmp[ch])
        ps = model.predict_proba(df_tmp[covariates])[:,1]
        raw_w = np.where(df_tmp[ch]==1, 1/ps, 1/(1-ps))
        df_tmp[f'ipw_{ch}'] = np.clip(raw_w, None, clip_upper) if clip_upper else raw_w

    def get_weight(row):
        touched = [ch for ch in channels if row[ch] == 1]
        return np.mean([row[f'ipw_{ch}'] for ch in touched]) if touched else 1.0
    df_tmp['combined_weight'] = df_tmp.apply(get_weight, axis=1)

    cs = df_tmp.groupby('touch_set').apply(
        lambda g: pd.Series({
            'n_users': g['combined_weight'].sum(),
            'n_conv': (g['combined_weight'] * g['converted']).sum()
        })
    ).reset_index()

    sv = shapley_generic(channels, cs, v_count)
    total = sum(sv.values())
    return {ch: sv[ch]/total for ch in channels}

clip_values = [5,10,20,50,None]
sensitivity_results = {}

print(" Trimming 민감도 분석 중...\n")
for cv in clip_values:
    label = f"clip={cv}" if cv else "clip=없음"
    sensitivity_results[label] = run_ipw_shapley(df_clean, channels, covariates, clip_upper=cv)
    print(f" {label} 완료")

print(f"\n{'채널':<12}" + "".join([f"{k:<14}" for k in sensitivity_results.keys()]))
for ch in channels:
    row = f"{ch:<12}"
    for k in sensitivity_results.keys():
        row += f"{sensitivity_results[k][ch]:<14.1%}"
    print(row)
```

## 희소 조합 Shrinkage 적용

```python
def compute_shrunk_coalition_stats(df_clean, channels, k=20):
    # 가중치 적용된 기본 통계
    cs = df_clean.groupby('touch_set').apply(
        lambda g: pd.Series({
            'n_users': g['combined_weight'].sum(),
            'n_conv': (g['combined_weight'] * g['converted']).sum()
        })
    ).reset_index()

    global_rate = cs['n_conv'].sum() / cs['n_users'].sum()

    # Empirical Bayes shrinkage: 표본이 작을수록 전체평균 쪽으로 당김
    cs['raw_rate'] = cs['n_conv'] / cs['n_users']
    cs['shrunk_rate'] = (cs['n_conv'] + k * global_rate) / (cs['n_users'] + k)
    cs['n_conv_shrunk'] = cs['shrunk_rate'] * cs['n_users']

    return cs, global_rate

shrunk_cs, global_rate = compute_shrunk_coalition_stats(df_clean, channels, k=20)

print(f"전체 평균 전환율: {global_rate:.2%}\n")
print("소표본 조합의 shrinkage 효과 (표본 작은 순):")
small = shrunk_cs.nsmallest(8, 'n_users')[['touch_set', 'n_users','raw_rate','shrunk_rate']]
for _, r in small.iterrows():
    print(f" {set(r['touch_set'])}: n={r['n_users']:.0f}, {r['raw_rate']:.1%} -> {r['shrunk_rate']:.1%}")
```

## Shrinkage 적용한 Shapley 재계산 및 비교

```python
def v_count_shrunk(S, cs):
    S = frozenset(S)
    mask = cs['touch_set'].apply(lambda ts: ts.issubset(S))
    return cs[mask]['n_conv_shrunk'].sum()

print(" Shrinkage 적용 Shapley 계산 중...")
shap_shrunk = shapley_generic(channels, shrunk_cs, v_count_shrunk)
total_shrunk = sum(shap_shrunk.values())

print("\n===== 최종 비교: 원본 → IPW → IPW+Shrinkage =====")
print(f"{'채널':<12}{'원본':<12}{'IPW':<12}{'IPW+Shrink':<14}")
for ch in channels:
    orig = shap_count[ch]/total_count
    ipw = shap_ipw[ch]/total_ipw
    shrunk = shap_shrunk[ch]/total_shrunk
    print(f"{ch:<12}{orig:<12.1%}{ipw:<12.1%}{shrunk:<14.1%}")
```

## IPW 기반 ATE 추정 (채널별)

```python
def estimate_ate_ipw(df_clean, channel, covariates, clip_upper=10):
    df_tmp = df_clean.copy()

    #성향점수 추정
    model = LogisticRegression()
    model.fit(df_tmp[covariates], df_tmp[channel])
    ps = model.predict_proba(df_tmp[covariates])[:,1]

    treated = df_tmp[channel].values
    y = df_tmp['converted'].values

    # IPW 가중치 (trimming 적용)
    w = np.where(treated == 1, 1/ps, 1/(1-ps))
    w = np.clip(w, None, clip_upper)

    # IPW 기반 ATE: 가중 평균의 차이
    y1_hat = np.sum(w * treated * y) / np.sum(w * treated) # 노출됐을 때 기대 전환율
    y0_hat = np.sum(w * (1-treated) * y) / np.sum(w * (1-treated)) # 노출 안 됐을 때 기대 전환율

    ate = y1_hat - y0_hat

    # 비교용: 단순 차이 (보정 전)
    naive = y[treated==1].mean() - y[treated==0].mean()

    return {
        'channel': channel,
        'naive_diff': naive,
        'ate_ipw': ate,
        'y1_hat': y1_hat,
        'y0_hat': y0_hat
    }

print(" 채널별 ATE 추정 중...\n")
ate_results = []
for ch in channels:
    r = estimate_ate_ipw(df_clean, ch, covariates)
    ate_results.append(r)
    print(f"{ch}:")
    print(f" 단순 차이(보정 전): {r['naive_diff']:+.2%}p")
    print(f" IPW 보정 ATE:       {r['ate_ipw']:+.2%}p")
    print(f" (노출 시 {r['y1_hat']:.2%} vs 미노출 시 {r['y0_hat']:.2%})\n")

ate_df = pd.DataFrame(ate_results)
```

## ATE 부트스트랩 신뢰구간

```python
def bootstrap_ate(df_clean, channel, covariates, n_bootstrap=100, clip_upper=10):
    ates = []
    for _ in range(n_bootstrap):
        sample = df_clean.sample(frac=1, replace=True)
        try:
            r = estimate_ate_ipw(sample, channel, covariates, clip_upper)
            ates.append(r['ate_ipw'])
        except Excution:
            continue
    return ates

print(" ATE 부트스트랩 (각 채널 100회)...\n")
for ch in channels:
    ates = bootstrap_ate(df_clean, ch, covariates, n_bootstrap=100)
    mean = np.mean(ates)
    ci_low, ci_high = np.percentile(ates, [2.5, 97.5])
    is_sig = (ci_low > 0) or (ci_high < 0)
    print(f"{ch}: ATE {mean:+.2%}p, 95% CI [{ci_low:+.2%}p, {ci_high:+.2%}p] -> {'유의미' if is_sig else '유의미하지 않음'}")
```

## 채널별 ATE 추정 (처치군 vs 무노출 대조군)

```python
def estimate_ate_vs_control(df_all, channel, covariates, clip_upper=10):
    # 이 채널 노출자 + 전체 대조군만 사용 (다른 채널 노출자는 제외)
    sub = df_all[(df_all[channel] == 1) | (df_all['group'] == 'control')].copy()
    sub['treated'] = (sub[channel] == 1).astype(int)

    model = LogisticRegression()
    model.fit(sub[covariates], sub['treated'])
    ps = model.predict_proba(sub[covariates])[:, 1]

    treated = sub['treated'].values
    y = sub['converted'].values

    w = np.where(treated == 1, 1/ps, 1/(1-ps))
    w = np.clip(w, None, clip_upper)

    y1_hat = np.sum(w * treated * y) / np.sum(w * treated)
    y0_hat = np.sum(w * (1-treated) * y) / np.sum(w * (1-treated))
    ate = y1_hat - y0_hat

    naive = y[treated==1].mean() - y[treated==0].mean()

    return {'channel': channel, 'n_treated': treated.sum(), 'n_control': (1-treated).sum(),
            'naive_diff': naive, 'ate_ipw': ate, 'y1_hat': y1_hat, 'y0_hat': y0_hat}

print(" 채널별 ATE 추정 중 (대조군 = 무노출자)...\n")
ate_results_v2 = []
for ch in channels:
    r = estimate_ate_vs_control(df_all, ch, covariates)
    ate_results_v2.append(r)
    print(f"{ch} (노출 {r['n_treated']:.0f}명 vs 무노출 {r['n_control']:.0f}명):")
    print(f"  단순 차이:    {r['naive_diff']:+.2%}p")
    print(f"  IPW 보정 ATE: {r['ate_ipw']:+.2%}p")
    print(f"  (노출 시 {r['y1_hat']:.2%} vs 무노출 시 {r['y0_hat']:.2%})\n")
```

## 부트스트랩 신뢰구간

```python
def bootstrap_ate_v2(df_all, channel, covariates, n_bootstrap=100, clip_upper=10):
    sub = df_all[(df_all[channel] == 1) | (df_all['group'] == 'control')].copy()
    ates = []
    for _ in range(n_bootstrap):
        sample = sub.sample(frac=1, replace=True)
        try:
            sample['treated'] = (sample[channel] == 1).astype(int)
            model = LogisticRegression()
            model.fit(sample[covariates], sample['treated'])
            ps = model.predict_proba(sample[covariates])[:, 1]
            treated = sample['treated'].values
            y = sample['converted'].values
            w = np.clip(np.where(treated==1, 1/ps, 1/(1-ps)), None, clip_upper)
            y1 = np.sum(w*treated*y)/np.sum(w*treated)
            y0 = np.sum(w*(1-treated)*y)/np.sum(w*(1-treated))
            ates.append(y1-y0)
        except Exception:
            continue
    return ates

print(" ATE 부트스트랩 (각 채널 100회)...\n")
for ch in channels:
    ates = bootstrap_ate_v2(df_all, ch, covariates, n_bootstrap=100)
    mean = np.mean(ates)
    ci_low, ci_high = np.percentile(ates, [2.5, 97.5])
    is_sig = (ci_low > 0) or (ci_high < 0)
    print(f"{ch}: ATE {mean:+.2%}p, 95% CI [{ci_low:+.2%}p, {ci_high:+.2%}p] → {'유의미' if is_sig else '유의미하지 않음'}")
```

### 1) 채널별로 clip 값을 바꿔가며 ATE + 공변량 균형(SMD)을 한 번에 계산

```python
def diagnose_channel(df_all, channel, covariates, clip_values=[10, 20, 50, None]):
    sub = df_all[(df_all[channel] == 1) | (df_all['group'] == 'control')].copy()
    sub['treated'] = (sub[channel] == 1).astype(int)

    model = LogisticRegression()
    model.fit(sub[covariates], sub['treated'])
    ps = model.predict_proba(sub[covariates])[:, 1]

    treated = sub['treated'].values
    y = sub['converted'].values
    raw_diff = y[treated==1].mean() - y[treated==0].mean()

    rows = []
    for clip in clip_values:
        w = np.where(treated == 1, 1/ps, 1/(1-ps))
        w_c = np.clip(w, None, clip) if clip else w

        y1 = np.sum(w_c*treated*y) / np.sum(w_c*treated)
        y0 = np.sum(w_c*(1-treated)*y) / np.sum(w_c*(1-treated))

        smds = {}
        for cov in covariates:
            m1 = np.sum(w_c[treated==1]*sub.loc[treated==1, cov]) / np.sum(w_c[treated==1])
            m0 = np.sum(w_c[treated==0]*sub.loc[treated==0, cov]) / np.sum(w_c[treated==0])
            smds[cov] = (m1 - m0) / sub[cov].std()

        rows.append({
            'channel': channel, 'clip': clip, 'n_treated': treated.sum(),
            'raw_diff': raw_diff, 'ate': y1 - y0,
            'max_abs_smd': max(abs(v) for v in smds.values()),
            **{f'smd_{k}': v for k, v in smds.items()}
        })
    return pd.DataFrame(rows)

all_results = pd.concat([diagnose_channel(df_all, ch, covariates) for ch in channels], ignore_index=True)
pd.set_option('display.width', 140)
print(all_results.round(4))
```

```python
def diagnose_channel(df_all, channel, covariates, clip_values=[10, 20, 50, None]):
    sub = df_all[(df_all[channel] == 1) | (df_all['group'] == 'control')].copy()
    sub['treated'] = (sub[channel] == 1).astype(int)

    rows = []
    for clip_upper in clip_values:
        model = LogisticRegression()
        model.fit(sub[covariates], sub['treated'])
        ps = model.predict_proba(sub[covariates])[:, 1]
        treated = sub['treated'].values
        y = sub['converted'].values

        w = np.where(treated == 1, 1/ps, 1/(1-ps))
        w = np.clip(w, None, clip_upper) if clip_upper else w

        y1 = np.sum(w * treated * y) / np.sum(w * treated)
        y0 = np.sum(w * (1-treated) * y) / np.sum(w * (1-treated))
        ate = y1 - y0
        raw_diff = y[treated==1].mean() - y[treated==0].mean()

        row = {
            'channel': channel, 'clip': clip_upper,
            'n_treated': treated.sum(), 'raw_diff': raw_diff, 'ate': ate,
        }
        smds = {}
        for cov in covariates:
            wm_t = np.average(sub.loc[treated==1, cov], weights=w[treated==1])
            wm_c = np.average(sub.loc[treated==0, cov], weights=w[treated==0])
            smds[f'smd_{cov}'] = (wm_t - wm_c) / sub[cov].std()
        row['max_abs_smd'] = max(abs(v) for v in smds.values())
        row.update(smds)
        rows.append(row)
    return pd.DataFrame(rows)

covariates = ['is_mobile', 'log_prior_freq', 'has_past_order',
              'is_logged_in', 'has_checkout', 'has_cart_add']

all_results = pd.concat([diagnose_channel(df_all, ch, covariates) for ch in channels], ignore_index=True)
print(all_results.to_string(index=False))
```
